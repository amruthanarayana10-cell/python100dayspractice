```python
#Turtle Challenge-3: Draw different shapes

from ColabTurtlePlus.Turtle import *
import random

clearscreen()

tim = Turtle()

colors = ["deep pink","indigo","dark magenta","olive drab","cornflower blue","dark orchid","indian red","sea green"]

def draw_shape(num_sides):
  angle = 360 / num_sides
  for _ in range(num_sides):
    tim.forward(100)
    tim.right(angle)


for shape_side_n in range(3,11):
  tim.color(random.choice(colors))
  draw_shape(shape_side_n)
```

```python
#Turtle Challenge-4: Draw a Random Walk

from ColabTurtlePlus.Turtle import *
import random

clearscreen()

tim = Turtle()

colors = ["deep pink","indigo","dark magenta","olive drab","cornflower blue","dark orchid","indian red","sea green"]
directions = [0, 90, 180, 270]
tim.pensize(15) # pensize = width
tim.speed(10) # speed = "fast"

for _ in range(200):
  tim.color(random.choice(colors))
  tim.forward(30)
  tim.setheading(random.choice(directions))
```

```python
# Python Tuples & How to Generate Random RGB Colors

#Turtle Challenge: Draw a Random Walk

from ColabTurtlePlus.Turtle import *
import random

clearscreen()

tim = Turtle()
colormode(255)

def random_color():
  r = random.randint(0, 255)
  g = random.randint(0, 255)
  b = random.randint(0, 255)
  random_color = (r, g, b)
  return random_color
  #return f"#{r:0.2X}{g:0.2X}{b:0.2X}" # Hexa-decimal color string


# colors = ["deep pink","indigo","dark magenta","olive drab","cornflower blue","dark orchid","indian red","sea green"]
directions = [0, 90, 180, 270]
tim.pensize(15) # pensize = width
tim.speed(10) # speed = "fast"

for _ in range(200):
  tim.color(random_color())
  tim.forward(30)
  tim.setheading(random.choice(directions))
```

```python
# Turtle Challenge 5- Draw a Spirograph

from ColabTurtlePlus.Turtle import *

clearscreen()

import random



tim = Turtle()
colormode(255)

def random_color():
  r = random.randint(0, 255)
  g = random.randint(0, 255)
  b = random.randint(0, 255)
  color = (r, g, b)
  return color

tim.speed(10) # speed(10)= "fast"


def draw_spirograph(size_of_gap):
  for _ in range(int(360 / size_of_gap)):
    tim.color(random_color())
    tim.circle(100)
    tim.setheading(tim.heading() + size_of_gap)


draw_spirograph(5)
```

The Hirst Painting Project Part-2:

Drawing the Dots

```python
from ColabTurtlePlus.Turtle import *

clearscreen()

# turtle.colormode(255)
tim = Turtle()
tim.speed(10)
tim.penup()
tim.hideturtle()
color_list = [(245, 241, 243),(178, 12, 36), (229, 244, 237), (182, 76, 26), (213, 158, 78), (245, 214, 75), (178, 172, 16), (236, 241, 247), (173, 22, 15), (118, 183, 202), (220, 130, 160), (61, 50, 107), (53, 97, 156), (66, 36, 59), (159, 53, 75), (245, 61, 27), (37, 135, 90), (35, 53, 69), (103, 188, 161), (218, 72, 131), (78, 46, 38), (237, 203, 2), (21, 167, 149), (53, 172, 187), (150, 211, 221), (161, 183, 229), (44, 62, 61), (233, 166, 182), (80, 104, 184)]

tim.setheading(225)
tim.forward(300)
tim.setheading(0)
number_of_dots = 100

for dot_count in range(1, number_of_dots + 1):
  tim.dot(20, random.choice(color_list))
  tim.forward(50)

  if dot_count % 10 == 0:
    tim.setheading(90)
    tim.forward(50)
    tim.setheading(180)
    tim.forward(500)
    tim.setheading(0)
```
