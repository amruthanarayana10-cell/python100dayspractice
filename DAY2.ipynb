Tip calculator

```python
print("Welcome to the tip calculator!")
total_amt =float(input("Total bill amount?$"))
tax = int(input("What percentage tip would you like to give? 10,12,or 15?"))
persons_num = int(input("how many people to split the bill? "))
bill_per_head =(((total_amt*(tax/100))+total_amt)/persons_num)
print(f"Total amount =$ {total_amt} \ntax percentage = {tax} \nnumber of persons = {persons_num} \nEach person should pay =$ {bill_per_head}")
print("Bill per person",round(bill_per_head))
```
