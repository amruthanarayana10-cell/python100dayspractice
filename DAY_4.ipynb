```python
print("Welcome to the infamous game of Rock Paper Scissors!")
print("Your opposition is Mr.C")
print("Let's begin!")

game_images = [rock, paper, scissors]

import random

player =int(input("What do you choose? Type 0 for Rock, 1 for Paper or 2 for Scissors.\n"))
if player >=0 and player<=2:
  print(game_images[player])

computer_num = random.randint(0,2)
print("Mr.C choose:")
print(game_images[computer_num])

if player >=3 or player<0 :
  print("You have typed an invalid number. You lose!")
elif player ==0 and computer_num==2:
  print("You win!")
elif computer_num ==0 and player==2:
  print("You lose!")
elif computer_num > player:
  print("You lose!")
elif computer_num < player:
  print("You win!")
elif computer_num == player:
  print("It\"s a draw!")

```
