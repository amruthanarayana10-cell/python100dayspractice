```python
class User:
  def __init__(self,user_id, username):
    self.id = user_id
    self.username = username
    self.followers = 0
    self.following = 0

  def follow(self, user):
    user.followers += 1 # The user being followed gains a follower
    self.following += 1 # The current user (self) gains a 'following'



user_1 = User("001","amrutha")
user_2 = User("002","varshini")

user_1.follow(user_2)
print(f"user_1 followers: {user_1.followers}")
print(f"user_1 following: {user_1.following}")
print(f"user_2 followers: {user_2.followers}")
print(f"user_2 following: {user_2.following}")
```
